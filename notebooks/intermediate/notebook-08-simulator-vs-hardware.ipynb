## Imports and Purpose
This lab uses `pandas` to format the simulator-versus-hardware delta table and the recommendation output.

```python
import pandas as pd

sim = {"quality": 0.93, "runtime_s": 1.4, "decision_consistency": 0.95}
hw = {"quality": 0.85, "runtime_s": 6.8, "decision_consistency": 0.78}

rows = []
for k in sim:
    rows.append({"metric": k, "simulator": sim[k], "hardware": hw[k], "delta": hw[k] - sim[k]})

pd.DataFrame(rows)
```

```python
if hw["quality"] >= 0.80 and hw["decision_consistency"] >= 0.75:
    print("Recommendation: Candidate acceptable for iterative tuning.")
else:
    print("Recommendation: Revisit circuit and noise assumptions before next hardware cycle.")
```
