```python
import pandas as pd

runs = pd.DataFrame([
    {"config": "qaoa_depth1", "score": 0.71, "runtime_s": 1.7},
    {"config": "qaoa_depth1", "score": 0.73, "runtime_s": 1.8},
    {"config": "qaoa_depth2", "score": 0.77, "runtime_s": 2.8},
    {"config": "qaoa_depth2", "score": 0.74, "runtime_s": 2.7},
    {"config": "vqe_ansatz_a", "score": 0.69, "runtime_s": 1.6},
    {"config": "vqe_ansatz_a", "score": 0.70, "runtime_s": 1.5},
])
runs
```

```python
summary = runs.groupby("config").agg(
    best_score=("score", "max"),
    mean_score=("score", "mean"),
    std_score=("score", "std"),
    mean_runtime=("runtime_s", "mean"),
).reset_index()
summary["quality_per_second"] = summary["mean_score"] / summary["mean_runtime"]
summary.sort_values(["mean_score", "quality_per_second"], ascending=False)
```
