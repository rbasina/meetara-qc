## Real-World Scenario
**Scenario:** Device Health Monitoring Before a Demo

**Problem:** You need to know whether hardware noise makes today's results unreliable.
**Baseline:** Compare ideal and noisy scores across repeated runs.
**Metric to watch:** Mean degradation percent and run-to-run variance.
**Practical takeaway:** Delay high-stakes demos if degradation exceeds tolerance.

## Imports and Purpose
This lab uses `pandas` to compare ideal and noisy runs and summarize degradation across trials.

```python
import pandas as pd

ideal_scores = [0.97, 0.96, 0.965, 0.968, 0.962]
noisy_scores = [0.89, 0.87, 0.88, 0.86, 0.875]

def degradation_percent(ideal, noisy):
    return (ideal - noisy) / ideal if ideal else None
```

```python
rows = []
for i, (a, b) in enumerate(zip(ideal_scores, noisy_scores), start=1):
    rows.append({"run": i, "ideal": a, "noisy": b, "degradation_pct": degradation_percent(a, b)})

df = pd.DataFrame(rows)
df
```

```python
summary = {
    "ideal_mean": df["ideal"].mean(),
    "noisy_mean": df["noisy"].mean(),
    "mean_degradation_pct": df["degradation_pct"].mean()
}
pd.DataFrame([summary])
```
