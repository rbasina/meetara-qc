```python
import pandas as pd

data = [
    {"candidate": "A", "depth": 12, "width": 2, "shots": 1024, "runtime_s": 1.3, "quality": 0.88},
    {"candidate": "B", "depth": 18, "width": 2, "shots": 1024, "runtime_s": 2.1, "quality": 0.91},
    {"candidate": "C", "depth": 10, "width": 3, "shots": 2048, "runtime_s": 2.9, "quality": 0.90},
]
df = pd.DataFrame(data)
df["quality_per_second"] = df["quality"] / df["runtime_s"]
df["quality_per_1000_shots"] = df["quality"] / (df["shots"] / 1000)
df
```

```python
df.sort_values(["quality", "quality_per_second"], ascending=False)
```
