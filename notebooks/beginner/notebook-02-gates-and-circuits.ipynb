## Imports and Purpose
This lab uses `QuantumCircuit` to define gate sequences, `AerSimulator` to execute them locally, and `pandas` to format the pass or investigate table.

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
import pandas as pd

sim = AerSimulator()

def run_counts(qc, shots=1024):
    result = sim.run(qc, shots=shots).result()
    return result.get_counts()
```

```python
x_qc = QuantumCircuit(1, 1)
x_qc.x(0)
x_qc.measure(0, 0)

x_counts = run_counts(x_qc, shots=1024)
p1 = x_counts.get("1", 0) / 1024
x_depth = x_qc.depth()

print("X-circuit counts:", x_counts)
print("Depth:", x_depth)
print("p(1):", round(p1, 4))
```

```python
bell_qc = QuantumCircuit(2, 2)
bell_qc.h(0)
bell_qc.cx(0, 1)
bell_qc.measure([0, 1], [0, 1])

bell_counts = run_counts(bell_qc, shots=1024)
p00 = bell_counts.get("00", 0) / 1024
p11 = bell_counts.get("11", 0) / 1024
corr = p00 + p11
bell_depth = bell_qc.depth()

print("Bell counts:", bell_counts)
print("Depth:", bell_depth)
print("p(00)+p(11):", round(corr, 4))
```

```python
rows = [
    {"circuit": "x_flip", "depth": x_depth, "metric": "p(1)", "value": p1, "threshold": ">= 0.98", "decision": "PASS" if p1 >= 0.98 else "INVESTIGATE"},
    {"circuit": "bell", "depth": bell_depth, "metric": "p(00)+p(11)", "value": corr, "threshold": ">= 0.90", "decision": "PASS" if corr >= 0.90 else "INVESTIGATE"},
]

pd.DataFrame(rows)
```
