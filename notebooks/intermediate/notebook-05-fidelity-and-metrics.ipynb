## Imports and Purpose
This lab uses `math` for the fidelity-style calculation and `pandas` to tabulate the metric results.

```python
import math
import pandas as pd

expected = {"00": 0.5, "11": 0.5}
observed = {"00": 0.46, "11": 0.44, "01": 0.05, "10": 0.05}

def tvd(p, q):
    keys = set(p) | set(q)
    return 0.5 * sum(abs(p.get(k, 0) - q.get(k, 0)) for k in keys)

def overlap_fidelity(p, q):
    keys = set(p) | set(q)
    return sum(math.sqrt(p.get(k, 0) * q.get(k, 0)) for k in keys)
```

```python
score_tvd = tvd(expected, observed)
score_fid = overlap_fidelity(expected, observed)
threshold = 0.90
decision = "PASS" if score_fid >= threshold else "INVESTIGATE"

pd.DataFrame([
    {"metric": "tvd", "value": score_tvd, "target": "low"},
    {"metric": "overlap_fidelity", "value": score_fid, "target": ">= 0.90"},
    {"metric": "decision", "value": decision, "target": "PASS"},
])
```
