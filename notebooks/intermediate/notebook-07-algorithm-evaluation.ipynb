## Imports and Purpose
This lab uses `pandas` to organize candidate configurations, compute quality-per-second, and rank the results.

```python
import pandas as pd

data = [
    {"config": "grover_k1", "score": 0.72, "runtime_s": 1.2},
    {"config": "grover_k2", "score": 0.84, "runtime_s": 1.8},
    {"config": "var_a", "score": 0.67, "runtime_s": 1.5},
    {"config": "var_b", "score": 0.75, "runtime_s": 2.1},
]
df = pd.DataFrame(data)
df["quality_per_second"] = df["score"] / df["runtime_s"]
df
```

```python
ranked = df.sort_values(["score", "quality_per_second"], ascending=False).reset_index(drop=True)
ranked
```
